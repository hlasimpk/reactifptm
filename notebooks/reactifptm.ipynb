# reactifptm — actifPTM interface confidence

Upload a PAE file (`.json` / `.npz` / `.npy` / `.pkl`) and a structure file
(`.pdb` or mmCIF `.cif`) from an AlphaFold-Multimer prediction, and this
notebook will compute the actifPTM interface confidence score using the
[`reactifptm`](https://pypi.org/project/reactifptm/) package.

**Steps:**
1. Run the *Setup* cell to install the package.
2. Run the *Upload files* cell and pick your PAE file and structure file.
3. (Optional) adjust the contact threshold in the *Options* cell.
4. Run the *Compute score* cell to see the results and download a JSON report.


## 1. Setup

```python
#@title Install reactifptm
!pip install -q reactifptm
print("reactifptm installed.")
```

## 2. Upload files

```python
#@title Upload your PAE file and structure file
from google.colab import files

print("Select your PAE file (.json / .npz / .npy / .pkl):")
pae_upload = files.upload()
pae_path = next(iter(pae_upload))

print("\nSelect your structure file (.pdb or .cif):")
structure_upload = files.upload()
structure_path = next(iter(structure_upload))

print(f"\nPAE file: {pae_path}")
print(f"Structure file: {structure_path}")
```

## 3. Options

```python
#@title Scoring options
threshold = 8.0  #@param {type:"number"}
output_filename = "results.json"  #@param {type:"string"}
```

## 4. Compute score

```python
#@title Run reactifptm
from reactifptm import Reactifptm
import json

r = Reactifptm(pae_path, structure_path, threshold=threshold)
overall, pairwise = r.compute_reactifptm()
r.save_results(output_filename)

print(f"Overall reactifptm score: {overall:.3f}\n")
print("Pairwise scores:")
for pair, score in pairwise.items():
    print(f"  {pair}: {score:.3f}")

with open(output_filename) as f:
    print("\nFull results:")
    print(json.dumps(json.load(f), indent=2))
```

## 5. Download results

```python
#@title Download results.json
from google.colab import files
files.download(output_filename)
```
